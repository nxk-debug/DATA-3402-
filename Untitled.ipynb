## Define Project

### Challenge Description
This project tackles a real-world software engineering problem: predicting whether a software module contains defects before it causes failures in production. Rather than relying solely on manual code reviews, the goal is to train a machine learning classifier that assigns a defect-probability score to each software module, allowing development teams to prioritize high-risk components for testing and review. The evaluation metric for this Kaggle competition is the Area Under the ROC Curve (AUC), which rewards good probability ranking rather than just binary accuracy.

### Dataset Description
The dataset originates from the NASA Metrics Data Program and is part of the Kaggle Playground Series Season 3. The training set contains 101,763 rows, each representing one software module described by 21 numerical features (software complexity metrics). Features include McCabe's cyclomatic complexity, Halstead metrics (volume, difficulty, effort), lines-of-code counts, and branching statistics. The binary target column `defects` indicates whether a module was found to contain bugs. The dataset is highly imbalanced - roughly 77% of modules are defect-free and 23% are defective - and heavily right-skewed due to a small number of extremely large, complex modules.

## Data Loading and Initial Look

```python
#Libraries 
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    roc_auc_score, accuracy_score, classification_report,
    ConfusionMatrixDisplay, RocCurveDisplay
)

sns.set_theme(style='whitegrid', palette='muted')
plt.rcParams['figure.dpi'] = 100
print('Libraries loaded.')
```

```python
#Load Data 
train_df = pd.read_csv('train.csv')
test_df  = pd.read_csv('test.csv')

print(f"Training set shape : {train_df.shape}")
print(f"Test set shape     : {test_df.shape}")
print(f"\nMissing values in training set : {train_df.isnull().sum().sum()}")
print(f"Missing values in test set     : {test_df.isnull().sum().sum()}")

train_df.head()
```

```python
#Feature Analysis Table
# Outlier definition: value > Mean + 3 × Standard Deviation (per feature)
feature_info = []
feature_cols = [c for c in train_df.columns if c not in ('id', 'defects')]

for col in feature_cols:
    series = train_df[col]
    dtype  = 'Categorical' if series.dtype == 'object' else 'Numerical'
    
    if dtype == 'Numerical':
        val_range = f"{series.min()} – {series.max()}"
    else:
        val_range = ', '.join(series.unique().astype(str)[:5])
    
    missing  = series.isnull().sum()
    cutoff   = series.mean() + 3 * series.std()
    outliers = int((series > cutoff).sum())
    
    feature_info.append({
        'Feature'      : col,
        'Type'         : dtype,
        'Values/Range' : val_range,
        'Missing'      : missing,
        'Outliers (>μ+3σ)': outliers
    })

feature_table = pd.DataFrame(feature_info)
print("\nFeature Analysis Table")
print(feature_table.to_string(index=False))
```

```python
#Class Balance 
target_counts = train_df['defects'].value_counts()
target_pct    = train_df['defects'].value_counts(normalize=True)

print("Target encoding  : False = no defect (0), True = defect (1)")
print(f"\nClass counts:\n{target_counts}")
print(f"\nClass proportions:\n{target_pct.round(4)}")
print("\n→ The dataset is IMBALANCED: ~77% defect-free, ~23% defective.")

fig, ax = plt.subplots(figsize=(5, 4))
target_counts.plot(kind='bar', ax=ax, color=['steelblue', 'tomato'], edgecolor='black')
ax.set_title('Class Distribution (defects)')
ax.set_xlabel('Defect')
ax.set_ylabel('Count')
ax.set_xticklabels(['False (no defect)', 'True (defect)'], rotation=0)
for p in ax.patches:
    ax.annotate(f'{int(p.get_height()):,}', (p.get_x()+p.get_width()/2, p.get_height()),
                ha='center', va='bottom', fontsize=10)
plt.tight_layout()
plt.show()
```

## Data Visualization

For each feature we compare the histogram between the two classes (defective vs. non-defective). Features where the two distributions diverge the most are the most informative for the ML task.

```python
#Per-feature histograms by class (all 21 features)
n_cols = 3
n_rows = int(np.ceil(len(feature_cols) / n_cols))

fig, axes = plt.subplots(n_rows, n_cols, figsize=(18, n_rows * 3.5))
axes = axes.flatten()

for idx, col in enumerate(feature_cols):
    ax = axes[idx]
    for label, color in [(False, 'steelblue'), (True, 'tomato')]:
        subset = train_df.loc[train_df['defects'] == label, col]
        ax.hist(subset, bins=50, alpha=0.55, color=color,
                label=f'defect={label}', density=True)
    ax.set_title(col, fontsize=10)
    ax.set_xlabel('')
    ax.legend(fontsize=8)

# Hide unused subplots
for j in range(len(feature_cols), len(axes)):
    axes[j].set_visible(False)

plt.suptitle('Feature Distributions by Class (raw scale)', fontsize=14, y=1.01)
plt.tight_layout()
plt.show()
```

Most features such as loc, v(g), n, v, e, t, lOCode, total_Op, and total_Opnd show a clear separation between defective and non-defective modules, indicating that they carry useful predictive information. In general, defective modules tend to have higher complexity values, which supports the idea that larger and more complex code is more likely to contain bugs. However, some features like l (program level) and lOBlank show weaker separation between the two classes and are less informative on their own. Additionally, all features are highly right-skewed, which suggests that applying a log transformation is important to reduce skewness and improve model performance.

## Data Cleaning and Preparation for Machine Learning

**Cleaning steps:**
1. Remove duplicate rows.
2. Drop the `id` column (row identifier, carries no predictive signal).
3. All features are already numerical — no missing values to impute.

**Rescaling strategy:**  
All 21 features are numerical and right-skewed with heavy outliers. We apply a log1p transform (log(1 + x)) to compress the long tails, followed by StandardScaler to give zero mean and unit variance. This is important for distance-based models; for Random Forest it is less critical but still improves interpretability.

```python
#Cleaning
n_before = len(train_df)
train_df = train_df.drop_duplicates()
print(f"Rows before dedup : {n_before}")
print(f"Rows after dedup  : {len(train_df)}")
print(f"Duplicates removed: {n_before - len(train_df)}")

# Separate features and target
X_raw = train_df.drop(columns=['id', 'defects']).copy()
y     = train_df['defects'].astype(int)   # True→1, False→0

print(f"\nFeature matrix shape : {X_raw.shape}")
print(f"Target unique values : {sorted(y.unique())}")
```

```python
#One-hot encode categorical features (required step)
cat_cols = X_raw.select_dtypes(include='object').columns.tolist()
print(f"Categorical columns detected: {cat_cols if cat_cols else 'None — all features are numerical'}")

if cat_cols:
    X_raw = pd.get_dummies(X_raw, columns=cat_cols, drop_first=True)
    print(f"Shape after one-hot encoding: {X_raw.shape}")
else:
    print("No encoding needed.")
```

```python
#Log1p transform + StandardScaler
X_log = np.log1p(X_raw)

scaler  = StandardScaler()
X_clean = pd.DataFrame(
    scaler.fit_transform(X_log),
    columns=X_raw.columns
)

print("Rescaling applied: log1p to StandardScaler")
print(f"X_clean shape: {X_clean.shape}")
X_clean.describe().round(2)
```

```python
#Visualize before and after cleaning for 4 representative features
sample_features = ['loc', 'v(g)', 'v', 'e']

fig, axes = plt.subplots(2, 4, figsize=(18, 7))

for i, col in enumerate(sample_features):
    # Before
    axes[0, i].hist(X_raw[col], bins=60, color='steelblue', edgecolor='white', alpha=0.8)
    axes[0, i].set_title(f'BEFORE: {col}', fontsize=10)
    axes[0, i].set_ylabel('Count')

    # After
    axes[1, i].hist(X_clean[col], bins=60, color='seagreen', edgecolor='white', alpha=0.8)
    axes[1, i].set_title(f'AFTER (log1p + scaled): {col}', fontsize=10)
    axes[1, i].set_ylabel('Count')

plt.suptitle('Feature Distributions Before and After Cleaning & Rescaling', fontsize=13)
plt.tight_layout()
plt.show()
```

## Machine Learning

###  Problem Formulation

```python
#Final feature matrix check
print("Columns removed  : id (identifier), defects (target)")
print("Columns kept     :", list(X_clean.columns))
print(f"\nFinal X shape    : {X_clean.shape}")
print(f"Target encoding  : 0 = no defect, 1 = defect")
print(f"\nTarget distribution:")
print(y.value_counts().rename({0: 'No defect (0)', 1: 'Defect (1)'}))
```

```python
# Train / Validation / Test Split (70 / 15 / 15)
X_train, X_temp, y_train, y_temp = train_test_split(
    X_clean, y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

X_val, X_test_split, y_val, y_test_split = train_test_split(
    X_temp, y_temp,
    test_size=0.50,
    random_state=42,
    stratify=y_temp
)

print(f"Training set   : {X_train.shape[0]:>6,} rows  ({X_train.shape[0]/len(X_clean):.0%})")
print(f"Validation set : {X_val.shape[0]:>6,} rows  ({X_val.shape[0]/len(X_clean):.0%})")
print(f"Test set       : {X_test_split.shape[0]:>6,} rows  ({X_test_split.shape[0]/len(X_clean):.0%})")
```

### Train ML Algorithm

We use a Random Forest Classifier as our primary model. It naturally handles:
- Non-linear relationships between complexity metrics and defects
- Feature importance ranking out of the box
- Probability calibration needed for AUC scoring

We set `class_weight='balanced'` to compensate for the 77/23 class imbalance.

```python
#Train Random Forest
model = RandomForestClassifier(
    n_estimators=200,
    max_depth=None,
    class_weight='balanced',   # handles class imbalance
    random_state=42,
    n_jobs=-1
)

model.fit(X_train, y_train)
print("Model training complete.")
print(f"  n_estimators : {model.n_estimators}")
print(f"  class_weight : balanced (corrects 77/23 imbalance)")
```

### Evaluate Performance on Validation Sample

```python
#Validation metrics
y_val_probs = model.predict_proba(X_val)[:, 1]
y_val_pred  = (y_val_probs >= 0.5).astype(int)

auc = roc_auc_score(y_val, y_val_probs)
acc = accuracy_score(y_val, y_val_pred)

print(f"Validation AUC      : {auc:.4f}    Kaggle competition metric")
print(f"Validation Accuracy : {acc:.4f}")
print()
print(classification_report(y_val, y_val_pred,
                             target_names=['No Defect (0)', 'Defect (1)']))
```

```python
#ROC Curve + Confusion Matrix 
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

RocCurveDisplay.from_predictions(y_val, y_val_probs,
                                 name='Random Forest', ax=axes[0])
axes[0].plot([0, 1], [0, 1], 'k--', label='Random baseline')
axes[0].set_title(f'ROC Curve — Validation Set (AUC = {auc:.4f})')
axes[0].legend()

ConfusionMatrixDisplay.from_predictions(
    y_val, y_val_pred,
    display_labels=['No Defect', 'Defect'],
    cmap='Blues', ax=axes[1]
)
axes[1].set_title('Confusion Matrix — Validation Set')

plt.tight_layout()
plt.show()
```

```python
#Hold-out test set evaluation
y_test_probs = model.predict_proba(X_test_split)[:, 1]
y_test_pred  = (y_test_probs >= 0.5).astype(int)

test_auc = roc_auc_score(y_test_split, y_test_probs)
test_acc = accuracy_score(y_test_split, y_test_pred)

print(f"Hold-out Test AUC      : {test_auc:.4f}")
print(f"Hold-out Test Accuracy : {test_acc:.4f}")
print()
print(classification_report(y_test_split, y_test_pred,
                             target_names=['No Defect (0)', 'Defect (1)']))
```

```python
#Feature Importance 
importances = pd.Series(model.feature_importances_, index=X_clean.columns)
importances = importances.sort_values(ascending=False)

fig, ax = plt.subplots(figsize=(10, 6))
importances.plot(kind='bar', ax=ax, color='steelblue', edgecolor='black')
ax.set_title('Random Forest Feature Importances')
ax.set_ylabel('Importance')
ax.set_xlabel('Feature')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

print("Top 5 most important features:")
print(importances.head())
```

###  Apply ML to the Kaggle Challenge Test Set

```python
#Prepare Kaggle test set with identical preprocessing
test_ids      = test_df['id']
X_test_kaggle = test_df.drop(columns=['id']).copy()

#one-hot encode categorical columns 
cat_cols_test = X_test_kaggle.select_dtypes(include='object').columns.tolist()
if cat_cols_test:
    X_test_kaggle = pd.get_dummies(X_test_kaggle, columns=cat_cols_test, drop_first=True)

#align columns with training set
X_test_kaggle = X_test_kaggle.reindex(columns=X_raw.columns, fill_value=0)

#apply same log1p + StandardScaler used on training
X_test_log    = np.log1p(X_test_kaggle)
X_test_scaled = pd.DataFrame(
    scaler.transform(X_test_log),   
    columns=X_raw.columns
)

print(f"Kaggle test set shape after preprocessing: {X_test_scaled.shape}")

#predict defect probabilities
test_probs = model.predict_proba(X_test_scaled)[:, 1]

#create submission file
submission = pd.DataFrame({'id': test_ids, 'defects': test_probs})
submission.to_csv('submission.csv', index=False)

print("\nsubmission.csv created successfully!")
print(submission.head())
```

The dataset consists of 101,763 training samples with 21 numerical features and no missing values. The target variable, defects, is binary and imbalanced, with approximately 77% non-defective modules and 23% defective modules. All features contain outliers defined as values greater than the mean plus three standard deviations, which were addressed using a log1p transformation to reduce skewness. Since the dataset contains only numerical features, no categorical encoding was required. The preprocessing pipeline included log transformation followed by StandardScaler normalization. A Random Forest model with 200 trees and class_weight='balanced' was trained to handle class imbalance. The model achieved a validation ROC-AUC score of around 0.77, which is the primary evaluation metric for this Kaggle competition. The final submission file contains predicted defect probabilities for the test set. Key findings show that defective modules consistently have higher complexity across most metrics, with features such as cyclomatic complexity (v(g)), lines of code (loc), Halstead metrics (v, n), and branch count being the most predictive. While class imbalance was partially addressed using balanced class weights, further improvements could be achieved through techniques like SMOTE or threshold tuning to improve detection of defective modules.
